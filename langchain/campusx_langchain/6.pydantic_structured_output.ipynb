```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from dotenv import load_dotenv
import os

from langchain_openai import ChatOpenAI

# Review Summerizer, Sentiment-Analyzer using Pydantic Models
```

```python
# Define model
# Note that hugging face doesnt support pydantic schemas

# repo_id = "meta-llama/Llama-3.1-8B-Instruct"
# llm = HuggingFaceEndpoint(
#     repo_id= repo_id,
#     max_new_tokens= 80,
#     temperature= 0.5,
#     huggingfacehub_api_token= os.getenv("HUGGINGFACEHUB_API_TOKEN")
# )
# llm = ChatHuggingFace(llm= llm)

llm = ChatOpenAI(
    model="gpt-4.1-nano",
    max_completion_tokens= 80,
    temperature= 0.3,
    api_key= os.getenv("OPENAI_API_KEY")
)
```

```python
# Schema 
from pydantic import BaseModel, Field
from typing import Literal

class LLMOut(BaseModel):
    summery : str = Field(...,description="An overall summary of the review")
    
    sentiment : Literal["positive","negative"] = Field(...,
                        description="A sentiment analysis of review either positive or negative"
                        )
    
    ideas : list[str] = Field(...,
                        description="A list of few ideas described in the review"
                        )
```

```python
model = llm.with_structured_output(schema= LLMOut)
```

```python
text = "This place has hands down the best brunch in town! The avocado toast with poached eggs was incredible, and their freshly squeezed orange juice was the perfect complement. The restaurant has such a bright, cheerful atmosphere with friendly staff who make you feel right at home. I've been here three weekends in a row and plan to keep coming back. Don't miss their homemade pastries - they're to die for!"

response = model.invoke(text)
print(response)
```

```python
print(response.summary, response.sentiment)
```
