```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
# initialize the model

repo_id = "meta-llama/Llama-3.1-8B-Instruct"

llm = HuggingFaceEndpoint(
    repo_id= repo_id,
    max_new_tokens= 50,
    temperature= 0.5,
    huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_API_TOKEN")
)

llm = ChatHuggingFace(llm = llm)
```

```python
# Topic name 
topic = "NLP transformers"
```

```python
# Lets try usually without StrOutParser and chains

prompt1 = f"Prepare a detailed report based on this topic: {topic}"

response1 = llm.invoke(prompt1)

prompt2 = f"Prepare a 5 line summary based on this report : {response1.content}"

response2 = llm.invoke(prompt2)

print(response2.content)
```

```python
# Using Output Parser

from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
    # we will use chains
# llm is model, prompt1 prompt2 are prompts, we need parser

prompt1 = PromptTemplate(
    template="Prepare a detailed report on this topic : {topic} "
    , input_variables=["topic"]
)

prompt2 = PromptTemplate(
    template="Prepare a 5 line summary for this text : {text} "
    , input_variables=["text"]
)

parser = StrOutputParser() # initialize the parser object

chain = prompt1 | llm | parser | prompt2 | llm | parser

topic = "cyclones"
response = chain.invoke({"topic":topic})

print(response)
```
