```python
from pydantic import BaseModel, Field
```

```python
# Define Schema 
class Student(BaseModel):
    name : str = "sachin"
    grade : int = Field(default=6) 
    roll_no : int =  Field(...,description="Roll number of student")
```

```python
# Define dict or object that we need to work with
student1 = {"roll_no":45} # since we created default vals, lets test them
```

```python
# Lets create a pydantic model object of previous object
# student1 which follows Student rules

student_model = Student(**student1) 
        # We need to unpack the dict items into Model class to
        # to initialize the model object

print(student_model)
```

```python
student_model.name, student_model.grade, student_model.roll_no
```

```python
# Lets verify implicit Type coercion by Pydantic

student2 = {"roll_no":"34"} # we passed str instead of int

student_model = Student( **student2)

print(student_model)
```
