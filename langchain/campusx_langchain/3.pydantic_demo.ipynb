```python
from pydantic import BaseModel, Field
```

```python
# Define Schema 
class Student(BaseModel):
    name : str
    roll_no : int =  Field(...,description="Roll number of student") 
```

```python
# Define dict or object that we need to work with
student1 = {"name":"Sachin", "roll_no":122}
```

```python
# Lets create a pydantic model object of previous object
# student1 which follows Student rules

student_model = Student(**student1) 
        # We need to unpack the dict items into Model class to
        # to initialize the model object

print(student_model)
```
