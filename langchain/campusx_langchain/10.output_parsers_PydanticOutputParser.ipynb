```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
# initialize the model

repo_id = "meta-llama/Llama-3.1-8B-Instruct"

llm = HuggingFaceEndpoint(
    repo_id= repo_id,
    max_new_tokens= 50,
    temperature= 0.5,
    huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_API_TOKEN")
)

llm = ChatHuggingFace(llm = llm)
```

```python
# Topic name 
topic = "NLP transformers"
```

```python
# Define Output schema
from pydantic import BaseModel, Field

class LLMOutput(BaseModel):
    name : str = Field(...,description="name of the person")
    age : int = Field(...,description="age", gt=18)
    city : str = Field(..., descrition="city of the person")
```

```python
# Using Structured Output Parser

from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate

# llm is model, we need parser, but we need to define response
# schemas first


parser = PydanticOutputParser(pydantic_object= LLMOutput)

# initialize the parser object

template = PromptTemplate(
    template="Give me name, age and city of a fictional person in this format : {format_instructions} "
    , input_variables=[],
    partial_variables= {"format_instructions":parser.get_format_instructions()}
)
# prompt = template.invoke({})

chain = template | llm | parser 

response = chain.invoke({}) # For invoke, 'input' is compulsory, hence just pass dummy {}

print(response)
```
