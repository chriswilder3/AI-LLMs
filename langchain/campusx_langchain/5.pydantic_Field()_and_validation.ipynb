```python
from pydantic import Field, BaseModel
```

```python
# Demonstrate pydantic elipsis and constraints

class Student(BaseModel):
    name : str = Field(...,description= "Name of student" )
            # 1st arg is ...(elipsis) Hence required field
    cgpa : float = Field(default=0,ge=0, le=10)
            # default = 0, and cgpa must be >= 0 & <= 10
```

```python
new_student = Student(**{"name" : "Sachin",
                       "cgpa":12})
```

It gives validation error : 

ValidationError: 1 validation error for Student
cgpa
  Input should be less than or equal to 10 [type=less_than_equal, input_value=12, input_type=int]

```python
# Lets correct it
new_student = Student(**{"name" : "Sachin",
                       "cgpa":8})

print(new_student)
```
