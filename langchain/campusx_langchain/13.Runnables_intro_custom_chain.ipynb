```python
import random


# Build custom dummy components, chains from scratch
```

```python
# Note that llm comp.s have predict method also
class DummyLLM:
    def __init__(self) -> None:
        print("LLM Created")
    
    def predict(self, prompt):
        response_list = ["Delhi is capital of India",
                         "Sachin is my name",
                         "Sky has darkish blue color"]
        
        return {"response":random.choice(response_list)}
    
```

```python
# Prompt template, note that they have format() method
class DummyPromptTemplate:
    def __init__(self, template, input_variables) -> None:
        self.template = template
        self.input_variables = input_variables

    def format(self, input_dict):
        # self.template is actually a sort of f-string
        # passed, hence we need to format it by passing input_dict
        # contents which are passed in template.format()
        return self.template.format(**input_dict)
```

```python
# Lets test it

llm = DummyLLM()

template = DummyPromptTemplate(
    template= "Write a {length} poem about {topic}",
    input_variables= ["length", "topic"]
)
```

```python
prompt = template.format({"length":"short","topic":"India"})

response = llm.predict(prompt)

print(response)
```

```python
# Now lets build chain class to allow users to use chains
# instead of custom stuff above
# It will have run() 
class DummyLLMChain:
    def __init__(self, llm, prompt) -> None:
        self.llm = llm
        self.prompt = prompt
    
    def run(self, input_dict):
        prompt = self.prompt.format(input_dict)
        return self.llm.predict(prompt)
```

```python
chain = DummyLLMChain(llm, template)
response = chain.run({"length":"short","topic":"India"})
```
