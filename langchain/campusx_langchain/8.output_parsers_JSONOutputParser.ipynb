```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
# initialize the model

repo_id = "meta-llama/Llama-3.1-8B-Instruct"

llm = HuggingFaceEndpoint(
    repo_id= repo_id,
    max_new_tokens= 50,
    temperature= 0.5,
    huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_API_TOKEN")
)

llm = ChatHuggingFace(llm = llm)
```

```python
# Topic name 
topic = "NLP transformers"
```

```python
# Using Json Output Parser

from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate
    # we will use chains
# llm is model, prompt is prompt, we need parser

parser = JsonOutputParser() # initialize the parser object

template = PromptTemplate(
    template="Give me name, age and city of a fictional person, in following format : {format_instructions} "
    , input_variables=[],
    partial_variables= {"format_instructions":parser.get_format_instructions()}
)
# prompt = template.invoke({})

chain = template | llm | parser 

response = chain.invoke({}) # For invoke, 'input' is compulsory, hence just pass dummy {}

print(response)
```

```python
# We can also do the same as follows

prompt = template.invoke({})
response = llm.invoke(prompt)
json_response = parser.invoke(response)
print(json_response)
```

```python
# Since its a dictionary we can access its contents

print(json_response["name"], type(json_response["name"]))
print(json_response["age"], type(json_response["age"]))
```
