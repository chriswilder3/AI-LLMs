```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
# initialize the model

repo_id = "meta-llama/Llama-3.1-8B-Instruct"

llm = HuggingFaceEndpoint(
    repo_id= repo_id,
    max_new_tokens= 50,
    temperature= 0.5,
    huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_API_TOKEN")
)

llm = ChatHuggingFace(llm = llm)
```

```python
# Topic name 
topic = "NLP transformers"
```

```python
# Using Structured Output Parser

from langchain_core.output_parsers import JsonOutputParser
from langchain.output_parsers import StructuredOutputParser, ResponseSchema
from langchain_core.prompts import PromptTemplate

# llm is model, we need parser, but we need to define response
# schemas first

schemas = [
    ResponseSchema(name="fact1", description="fact 1 about topic"),
    ResponseSchema(name="fact2", description="fact 2 about topic"),
    ResponseSchema(name="fact3", description="fact 3 about topic")
]

parser = StructuredOutputParser.from_response_schemas(
    schemas
) 

# initialize the parser object

template = PromptTemplate(
    template="Give 3 facts about {topic} in this format : {format_instructions} "
    , input_variables=['topic'],
    partial_variables= {"format_instructions":parser.get_format_instructions()}
)
# prompt = template.invoke({})

chain = template | llm | parser 

topic = "blackholes"
response = chain.invoke({"topic":topic}) # For invoke, 'input' is compulsory, hence just pass dummy {}

print(response)
```
